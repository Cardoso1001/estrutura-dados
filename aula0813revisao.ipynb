Usuário adicionando valores a lista


```python
alunos = []
qtd = int(input('Quantos aluos deseja cadastrar? '))

for i in range(qtd):
    nome = input(f'Digite o nome do aluno nº {i+1}: ')
    idade = int(input(f'Digite a idade do aluno nº {i+1}: '))
    alunos.append([nome, idade])

print("\nLista de alunos:")

for aluno in alunos:
  print(f"Aluno: {aluno[0]}. Idade: {aluno[1]}")

idade_filtro = int(input("\nMostrar alunos com idade maior ou igual a: "))

print(f"\nAlunos com idade maior ou igual a {idade_filtro}:")

for aluno in alunos:
  if aluno[1] >= idade_filtro:
    print(f"Aluno: {aluno[0]}. Idade: {aluno[1]}")
```

## Atv Menu de crud e filtros

```python
def insert():
  qtd = int(input("Quantos jogos deseja cadastrar? "))
  for i in range(qtd):
    nome = input("Digite o nome do jogo: ")
    preco = float(input("Digite o preço do jogo: "))
    jogos.append([nome, preco])
    print("Jogo cadastrado com sucesso!")

# Fim de insert

def show():
  print("\nLista de jogos:")
  for jogo in jogos:
    print(f"Jogo: {jogo[0]}. Preço: {jogo[1]}")

# Fim de show

def update():
  nome = input("Digite o nome do jogo que deseja alterar o preço: ")
  for jogo in jogos:
    if jogo[0] == nome:
      novo_preco = float(input("Digite o novo preço do jogo: "))
      jogo[1] = novo_preco

# Fim de update

def delete():
  nome = input("Digite o nome do jogo que deseja excluir: ")
  for jogo in jogos:
    if jogo[0] == nome:
      jogos.remove(jogo)
      print("Jogo excluído com sucesso!")
      return

# Fim de delete

def filter():
  preco_filtro = float(input("Digite o preço máximo dos jogos que deseja listar: "))
  print(f"\nJogos com preço menor ou igual a {preco_filtro}:")
  for jogo in jogos:
    if jogo[1] <= preco_filtro:
      print(f"Jogo: {jogo[0]}. Preço: {jogo[1]}")

# Fim de filtro

def main():
  print("---- JOGOS----\n\n")
  while True:
    # Escolhendo opções de menu
    print("\n\n --- MENU --- \n")
    print("Digite a opção:")
    print("1. Cadastrar Jogos")
    print("2. Listar jogos")
    print("3. Alterar jogo")
    print("4. Excluir jogo")
    print("5. Filtrar jogos")
    print("0. SAIR")

    opcao = int(input("\n ->> Digite sua opção: "))

    match opcao:
      case 0: #Sair
        print("\n-- VOCÊ SAIU! --")
        break
      case 1: #create
        insert()
      case 2: #read
        show()
      case 3: #update
        update()
      case 4: #delete
        delete()
      case 5: #filtrando valores de cada jogo, retornando apenas os mais baratos
        filter()
# Fim do Menu Principal

jogos = []

main()
```
