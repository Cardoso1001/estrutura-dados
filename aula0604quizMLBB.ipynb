```python
!pip install google.genai
import os
from google.colab import userdata
from google import genai
from google.genai import types
os.environ['CardososKey'] = userdata.get('CardososKey')
client = genai.Client(api_key = os.environ['CardososKey'])
model = "gemini-2.5-flash"

chat_config = types.GenerateContentConfig(system_instruction="Você é um jogo de quiz do MLBB. Você escolherá um personagem do jogo Mobile Legends: Bang Bang, e vai deixar o usuário tentar até 6 vezes acertar o personagem. Estas são as dicas: 1- Um paragrafo curto da história dele, sem dar dica de como ele é fisicamente nem habilidades; 2- O tipo de dano majoritário(mágico, fisico, ou hibrido caso ele use realmente os dois); 3- Classe do personagem, como mago, atirador, soldado, etc; 4- Genero do personagem; 5- Um personagem que interage com este; 6- A primeira letra do nome dele. Ele tem apenas 6 chances para tentar acertar o personagem, caso acerte, comemore com o usuário e já escolha outro personagem, caso erre, revele uma dica até as chances terminarem. No final das tentativaas, revele o personagem e gere o próximo quiz de personagem.")
chat = client.chats.create(model = model, config = chat_config)
```

```python
print("Bem-vindo ao MLBB QUIZ.")
print("Inicialmente, escolherei um herói aleatório do jogo, e você terá que adivinha-lo.")
print("Ao final, você deverá falar 'SAIR' para encerrar o quiz.")
prompt = "Vamos começar o jogo. Escolha o personagem!"

while prompt.lower() != "sair":
  response = chat.send_message(prompt)
  print(response.text)
  prompt = input("Adivinhe o herói: ")

print("Você saiu o chat.")
```
