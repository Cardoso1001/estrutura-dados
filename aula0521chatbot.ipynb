```python
# Para instalar coisas aqui no Colab
!pip install google.genai #google > IA GENerativa
import os #Manipular arquivos, editar, criar, diretórios
from google.colab import userdata # gerenciar uma chave api de forma segura
from google import genai
from google.genai import types # contém funções principais para comunicação com os modelos de IA

# Ir na chave ali na esquerda. Depois criar uma chave API Gemini caso não tenha criado.
# Caso tenha criado, colocar o nome, e a chave api  em valor
```

```python
# Chave api recebe 'segurança'
os.environ['CardososKey'] = userdata.get('CardososKey')

client = genai.Client(api_key = os.environ['CardososKey'])

# Passar agora que vamos usar o modelo gemini
model = "gemini-2.5-flash" # Gemini versão 2.5
```

```python
# Instruções à IA e como ela deve se comportar nesse chatbot
chat_config = types.GenerateContentConfig(system_instruction="Você é um assisteente que interage com o usuário apenas sobre assuntos relacionado a matemática. Seja rigoroso e sem piedade, faça com que o cliente aprenda matemática de qualquer forma, sempre perguntando se ele compreendeu sobre o assunto. Se ele mudar de assunto, passe um calculo simples para ele fazer em tom de ironia.")
```

```python
chat = client.chats.create(model = model, config = chat_config)
```

```python
print("Bem-vindo, sou o seu professor de matemática.")
print("Farei com que você entenda matemática de um jeito ou de outro.")
print("Ao final, você deverá falar 'APRENDI' para encerrar a aula.")
prompt = input("Digite sua pergunta: ")

while prompt.lower() != "aprendi":
  response = chat.send_message(prompt)
  print(response.text)
  prompt = input("Digite sua pergunta: ")

print("De nada, aliás.")
```

```python
print("ESCALAÇÃO DO TIME")
# Criando a lista

jogadores = ["Leo Pereira", "Vini Jr", "Casimiro", "Alisson","Endrick"]

print("\nLista original: ")
print(jogadores)

# acessando posições
print("\nPrimeiro jogador: ")
print(jogadores[0])

print("\nÚltimo jogador: ")
print(jogadores[-1])

print("\nSubstituindo jogadores...")
jogadores[1] = "L. Henrique"
jogadores[-1] = "Lucas Pquetá"

print("\nLista tualizada: ")
print(jogadores)

# Adicionando Players
jogadores.append("Marquinhos")
print(jogadores)

jogadores.pop(2)
jogadores.remove("Marquinhos")
print(jogadores)
jogadores.sort()
print(jogadores)

posicao = int(input("Digite uma posição da lista: "))

if posicao >= 0 and posicao < len(jogadores):
  print(f"Jogador selecionado: num {posicao}. Jogador {jogadores[posicao]}")
else:
  print("Posião invalida!")

print("FIM DA ATIVIDADE")
```

```python
pessoas = []

print("Digite 10 nomes")
for p in range(10):
  nome = input(f"Nome n{p+1}: ")

  if nome in pessoas:
    print("Nome repetido!")
    print(f"Nomes: {pessoas}")

    while nome in pessoas:
      nome = input("Digite um outro nome que não está na lista: ")

  pessoas.append(nome)
  print(f"Nomes: {pessoas}")


print(f"Nomes: {pessoas}")
pessoas.sort()
print(f"Nomes em ordem alfabética: {pessoas}")
```
