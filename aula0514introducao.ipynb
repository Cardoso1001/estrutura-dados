# Desafio 2 - CRUD
Em aula fomos até inserir e listar.
Como tarefa de casa fiz a edição e exclusão dos itens.

```python
produtos = [
    {"nome":'Refrigerente',
    "preco":7.5,
    "quantidade":25},
    {"nome":'Salgadinho',
    "preco":3.5,
    "quantidade":32},
    {"nome":'Skol Lata',
    "preco":4.5,
    "quantidade":56},
    {"nome":'Sorvetinho',
    "preco":0.5,
    "quantidade":120}
]

def cadastrar_produtos():
  nome = input("Digite o nome do produto: ")
  preco = float(input("Digite o preço do produto: "))
  quantidade = int(input("Digite a qunatidade do produto: "))

  produto = {
    "nome":nome,
    "preco":preco,
    "quantidade":quantidade
  }

  produtos.append(produto)
  print(" >>> Produto cadastrado com sucesso!")
# Fim de cadastrar_produtos

def listar_produtos():
  if len(produtos) == 0:
    print("Nenhum produto encontrado.\n")
    return
  print("\n--- LISTA DE PRODUTOS ---\n")

  # for produto in produtos:
  #   print(f"Nome: {produto['nome']}")
  #   print(f"Preço: R$ {produto['preco']:.2f}")
  #   print(f"Quantidade: {produto['quantidade']}")
  #   print("\n---------------------------------\n")

  for id in range(len(produtos)):
    print(f"ID: {id+1}")
    print(f"Nome: {produtos[id]['nome']}")
    print(f"Preço: R$ {produtos[id]['preco']:.2f}")
    print(f"Quantidade: {produtos[id]['quantidade']}")
    print("\n---------------------------------\n")
# Fim de listar_produtos

def calcular_total_estoque():
  total = 0
  for produto in produtos:
    total += (produto["preco"] * produto["quantidade"])

  print("\n--------------------------------\n")
  print(f"O valor total do estoque é de R${total:.2f}")
  print("\n---------------------------------\n")
# Fim de calcular_total_estoque

def editar_produto():
  produto_novo = {}
  listar_produtos()
  selecao = int(input("Digite o número do produto para editar:"))

  print(f"\n--- EDITANDO PRODUTO ID {selecao} ---")
  print("(Digite 'CANCELAR' caso não queira editar o valor)\n\n")
  selecao -= 1

  print(f"Nome antigo: {produtos[selecao]["nome"]}")
  produto_novo['nome'] = input("Nome novo: ")
  print(f"Nome: {produto_novo['nome']}")
  print(f"Preço antigo: R$ {produtos[selecao]["preco"]:.2f}")
  produto_novo['preco'] = input("Preço novo: ")
  print(f"Preço: R$ {produto_novo['preco']}")
  print(f"Quantidade antiga: {produtos[selecao]["quantidade"]}")
  produto_novo['quantidade'] = input("Quantidade nova: ")
  print(f"Quantidade: {produto_novo['quantidade']}")

  print("\n---------------------------------\n")

  # Erro cabuloso quando tento cancelar um produto, não sei o porque
  if produto_novo['nome'].lower != 'cancelar' :
    produtos[selecao]['nome'] = produto_novo['nome']
  else:
    print("Nome não alterado!")
  if produto_novo['preco'].lower != 'cancelar' :
    produtos[selecao]['preco'] = float(produto_novo['preco'])
  else:
    print("Preço não alterado!")
  if produto_novo['quantidade'].lower != 'cancelar' :
    produtos[selecao]['quantidade'] = int(produto_novo['quantidade'])
  else:
    print("Quantidade não alterado!")

  # Operador ternário
  # produtos[selecao]['nome'] = produto_novo['nome'] if "cancelar" != produto_novo['nome'].lower else print("Nome não alterado!")
  # produtos[selecao]['preco'] = float(produto_novo['preco']) if "cancelar" != produto_novo['preco'].lower else print("Preço não alterado!")
  # produtos[selecao]['quantidade'] = int(produto_novo['quantidade']) if "cancelar" != produto_novo['quantidade'].lower else print("Quantidade não alterado!")

  print(f"ID: {selecao+1}")
  print(f"Nome: {produtos[selecao]['nome']}")
  print(f"Preço: R$ {produtos[selecao]['preco']:.2f}")
  print(f"Quantidade: {produtos[selecao]['quantidade']}")
  print("\n---------------------------------\n")

#Fim de editar_produto

def excluir_produto():
  listar_produtos()
  selecao = int(input("Digite o número do produto para excluir:"))
  selecao -= 1

  print(f"ID: {selecao+1}")
  print(f"Nome: {produtos[selecao]['nome']}")
  print(f"Preço: R$ {produtos[selecao]['preco']:.2f}")
  print(f"Quantidade: {produtos[selecao]['quantidade']}")
  print("\n---------------------------------\n")

  print("\n--- DESEJA EXCLUIR ESSE PRODUTO? ---\n")
  while True:
    print("1. Sim")
    print("0. Não")
    resp = int(input("Resposta: "))

    if resp == 1:
      produtos.pop(selecao)
      print("Produto excluído!")
      return
    elif resp == 0:
      print("Operação cancelada. Retornando ao menu.")
      return
    print("Opção inválida. Digite novamente.\n")
# Fim de excluir_produto

# Menu Principal
while True:
  print("\n--- SISTEMA DE PRODUTOS ---\n")
  print("1. Cadastrar produto")
  print("2. Listar produtos")
  print("3. Calcular valor total")
  print("4. Editar produto")
  print("5. Excluir produto")
  print("0. Sair\n")

  opcao = int(input("Digite a opção: "))
  print("\n")

  match opcao:
    case 0:
      print(" >>> Você Saiu.")
      break
    case 1:
      cadastrar_produtos()
    case 2:
      listar_produtos()
    case 3:
      calcular_total_estoque()
    case 4:
      editar_produto()
    case 5:
      excluir_produto()
    case _:
      print(" >>> Opção invalida!")
# Fim do Menu Principal
```
