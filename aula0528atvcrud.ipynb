## Func Cadastrar

```python
# Cadastrar tripulações
def cadastrar():
    print("-- CADASTRANDO TRIPULAÇÃO --\n")
    nome_capitao = input("Digite o nome do capitão: ")
    valor_capitao = float(input("Digite a recompensa do capitão: "))
    qtde = int(input(f"Quantos piratas a tripulação de {nome_capitao} possui?\n "))

    # 1. Cria a estrutura base da nova tripulação
    nova_tripulacao = {
        "nome_capitao": nome_capitao,
        "valor_capitao": valor_capitao,
        "piratas": []  # Inicializa a lista de piratas vazia
    }

    # 2. Coleta os dados de cada pirata e adiciona direto na lista
    for i in range(qtde):
        print(f"\n--- {i+1}º Pirata ---")
        nome_p = input("Digite o nome: ")
        valor_p = float(input("Digite a recompensa: "))

        # Cria o dicionário do pirata atual
        pirata = {
            "nome_pirata": nome_p,
            "valor_pirata": valor_p
        }
        # Adiciona o pirata na lista da nova tripulação
        nova_tripulacao["piratas"].append(pirata)

    # 3. Adiciona a tripulação completa na lista global
    tripulacoes.append(nova_tripulacao)

    print("\n-- TRIPULAÇÃO CADASTRADA --\n")
# fim de cadastrar
```

## Func Excluir

```python
# Excluir Tripulação
def excluir():
  print("\n -- EXCLUIR TRIPULAÇÃO -- ")
  # Imprimindo Tripulações
  for i in range(len(tripulacoes)):
    print(f"Tripulação número {i+1}. Capitão: {tripulacoes[i]["nome_capitao"]}")

  while True:
    # Escolhendo a tripulação
    print("Digite o número da tripulação, ou zero(0) para voltar para o menu.")
    opcao = int(input("Opção: "))

    # Saindo para menu
    if opcao == 0:
      print("\nVocê voltou ao menu...")
      return

    # Opção inválida
    elif opcao > len(tripulacoes) or opcao < 0:
      print("Opção inválida. Tente novamente.")

    # Ok - Excluindo tripulação.
    else:
      print(f"Tripulação de {tripulacoes[opcao-1]["nome_capitao"]} excluída.")
      tripulacoes.pop(opcao-1)
      return
# fim de excluir
```

## Ranking Top 5 $$$

```python
# TOP 5 Tripulações com maior recompensa
def top_recompensa():
  ordem = []
  # Pegando tripulação para 'cap'
  for cap in tripulacoes:
    # Somando valor de capitão tambem
    valor = cap["valor_capitao"]
    # Pegando piratas para 'trip'
    for trip in cap["piratas"]:
      # Somando valores de piratas
      valor = valor + (trip["valor_pirata"])
    # Add valor a lista criada para ordenar
    ordem.append({"capitao":cap["nome_capitao"], "total_recompensa":valor})
  # Ordenando de acordo com a recompensa e de forma decrescente
  ordem.sort(reverse = True, key = lambda x: x["total_recompensa"])

  print("--- COLOCAÇÃO DAS TRIPULAÇÕES COM MAIORES RECOMPENSAS ---\n\n")
  # Imprimindo Ranking TOP 5 $$$
  for i in range(5):
    print(f"Nº {i+1} - tripulação de {ordem[i]["capitao"]}: $$$ {ordem[i]["total_recompensa"]:.2f}")

# fim de top 5 recompensa
```

## Ranking Top 5 qtde

```python
# TOP 5 Tripulações com mais piratas
def top_qtde_piratas():
  ordem = []
  # Pegando tripulações para 'cap'
  for cap in tripulacoes:
    # Adicionando o capitão e o número de piratas à soma
    valor = 1 + len(cap["piratas"])
    # Add valor a lista criada para ordenar
    ordem.append({"capitao":cap["nome_capitao"], "n_piratas":valor})
  # Ordenando de acordo com a qunatidade de piratas e de forma decrescente
  ordem.sort(reverse = True, key = lambda x: x["n_piratas"])

  print("--- COLOCAÇÃO DAS TRIPULAÇÕES COM MAIOR NÚMERO DE PIRATAS ---\n\n")
  # Imprimindo Ranking TOP 5 qtde
  for i in range(5):
    print(f"Nº {i+1} - tripulação de {ordem[i]["capitao"]}: {ordem[i]["n_piratas"]} piratas.")

# fim de top 5 mais piratas
```

## Func Menu Principal

```python
# Menu Principal
def menu_principal():
  print("---- CAPITÃES E TRIPULAÇÕES DE PIRATAs DE ONE PIECE----\n\n")
  while True:
    # Escolhendo opções de menu
    print("\n\n --- MENU --- \n")
    print("Digite a opção:")
    print("1. Cadastrar Tripulação")
    print("2. Listar Tripuações")
    print("3. Alterar Tripulação de Capitão")
    print("4. Excluir Tripulação")
    print("5. Ver TOP 5 Tripulações com maior recompensa")
    print("6. Ver TOP 5 Tripulações com mais piratas")
    print("0. SAIR")

    opcao = int(input("\n ->> Digite sua opção: "))

    match opcao:
      case 0: #Sair
        print("\n-- VOCÊ SAIU! --")
        break
      case 1: #create
        cadastrar()
      case 2: #read
        listar()
      case 3: #update
        alterar()
      case 4: #delete
        excluir()
      case 5: #somando valores de cada tripulação e ordenando de maior para menor
        top_recompensa()
      case 6: #somando quantidade de cada tripulação e ordenando de maior para menor
        top_qtde_piratas()
# Fim do Menu Principal
```

## Executando Código

```python
# Executando código.
menu_principal()
```
